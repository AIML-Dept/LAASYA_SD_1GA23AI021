# **Theory**

 A qubit is the fundamental unit of quantum information. Unlike a classical bit that can store only 0 or 1, a qubit can exist in a combination of both states simultaneously, known as superposition.

 Quantum states are represented mathematically using Dirac (bra-ket) notation and state vectors. A single qubit in state |0⟩ is represented as [1, 0], while |1⟩ is represented as [0, 1].

#Quantum gates change the state of qubits. The X gate flips a qubit from |0⟩ to |1⟩, while the Hadamard (H) gate creates an equal superposition of |0⟩ and |1⟩.

 The Bloch sphere is a graphical representation of a single qubit state, whereas the Q-sphere is used to visualize multi-qubit quantum states. For every valid quantum state, the sum of the squares of all probability amplitudes must always equal 1, which is known as the normalization condition.
    

# **Exercise 1 (Easy)**

**Create a qubit in state |0⟩, apply an X gate, and print the resulting statevector.**

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_state_city
import matplotlib.pyplot as plt

# Create a quantum circuit with 1 qubit
qc = QuantumCircuit(1)

# Qubit starts in |0⟩
# Apply X gate
qc.x(0)

# Draw the circuit
qc.draw("mpl")
plt.show()

# Generate the statevector
state = Statevector.from_instruction(qc)

# Print the statevector
print("Statevector:")
print(state)

# Visualize the statevector
from IPython.display import display

display(qc.draw("mpl"))

display(plot_state_city(state))
```

 **Question: Create a qubit in superposition using the Hadamard (H) gate, then apply the S gate and visualize the state on the Bloch Sphere.**

```python
import sys
import subprocess
subprocess.check_call([sys.executable, "-m", "pip", "install", "qiskit", "matplotlib", "pylatexenc"])
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
from IPython.display import display

# Create a quantum circuit with 1 qubit
qc = QuantumCircuit(1)

# Apply Hadamard gate to create superposition
qc.h(0)

# Apply S gate (Phase gate)
qc.s(0)

# Display the quantum circuit
display(qc.draw("mpl"))

# Generate the statevector
state = Statevector.from_instruction(qc)

# Print the statevector
print("Statevector:")
print(state)

# Display the Bloch Sphere
display(plot_bloch_multivector(state))
```

## Observation

The Hadamard (H) gate places the qubit into an equal superposition of |0⟩ and |1⟩. The S gate adds a phase of i to the |1⟩ component without changing the probabilities of measuring 0 or 1. The Bloch Sphere shows that the qubit rotates around the Z-axis due to the phase shift introduced by the S gate.

## Conclusion

The experiment demonstrates that quantum gates can modify both the probability amplitudes and the phase of a qubit. The Hadamard gate creates superposition, while the S gate changes only the phase, which is clearly visualized using the Bloch Sphere.

# **Exercise 3 (Hard)**

**Question: Create a 2-qubit equal superposition state and verify that the sum of the probabilities is equal to 1 (normalization condition).**

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_state_qsphere
from IPython.display import display
import numpy as np

# Create a quantum circuit with 2 qubits
qc = QuantumCircuit(2)

# Apply Hadamard gate to both qubits
qc.h(0)
qc.h(1)

# Display the circuit
display(qc.draw("mpl"))

# Generate the statevector
state = Statevector.from_instruction(qc)

# Print the statevector
print("Statevector:")
print(state)

# Calculate probabilities
probabilities = np.abs(state.data) ** 2

print("\nProbabilities:")
print(probabilities)

print("\nSum of Probabilities =", np.sum(probabilities))

# Visualize the state on Q-Sphere
display(plot_state_qsphere(state))
```

## Observation

Applying the Hadamard gate to both qubits creates an equal superposition of the four basis states: |00⟩, |01⟩, |10⟩, and |11⟩. Each state has an equal probability of 0.25. The total probability is 1, confirming that the quantum state is normalized. The Q-Sphere visualization displays all four basis states with equal amplitudes.

## Conclusion

This experiment demonstrates that applying Hadamard gates to multiple qubits creates an equal superposition of all possible basis states. The probabilities of all states sum to 1, satisfying the normalization condition of quantum mechanics.

# **Exercise 4 (Real-World Application)**

**Question: Simulate a noisy quantum communication channel by applying a random phase error to a qubit and visualize the effect on the Bloch Sphere.**

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
from IPython.display import display
import numpy as np

# Create a quantum circuit with 1 qubit
qc = QuantumCircuit(1)

# Prepare the qubit in superposition
qc.h(0)

# Generate a random phase error
theta = np.random.uniform(0, 2 * np.pi)

# Apply the random phase error
qc.p(theta, 0)

# Display the quantum circuit
display(qc.draw("mpl"))

# Generate the statevector
state = Statevector.from_instruction(qc)

# Print the random phase angle
print(f"Random Phase Error (θ): {theta:.4f} radians")

# Print the resulting statevector
print("\nStatevector:")
print(state)

print("Amplitudes:")
print(state.data)

print("\nNormalization Check:")
print(sum(abs(state.data)**2))

# Visualize the qubit on the Bloch Sphere
display(plot_bloch_multivector(state))
```

## Observation

The qubit was first placed in a superposition state using the Hadamard gate. A random phase error was then introduced using the phase gate P(θ), simulating noise in a quantum communication channel. The Bloch Sphere visualization shows that the phase error changes the orientation of the qubit state while maintaining its quantum nature.

## Conclusion

This experiment demonstrates how a random phase error affects a qubit during quantum communication. The Bloch Sphere clearly shows the change in the qubit's state due to the applied phase error. Such simulations help in understanding the impact of noise on quantum systems and the need for quantum error correction.
